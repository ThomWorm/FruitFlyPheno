```python
# open hillsborough_testing_off_results.nc as xarray
import xarray as xr
import pandas as pd

data = xr.open_dataset("santaclara_testing_off_results.nc")
# convert days_to_completion for generation 3 to dates, using detection_date from attributes
# Convert days_to_completion for generation 3 to dates, using detection_date from attributes
detection_date = pd.to_datetime(data.attrs["detection_date"])
days_to_completion_gen3 = data["days_to_completion"].sel(generation=3)

# Ensure days_to_completion_gen3 is flattened to a 1D array
completion_dates_gen3 = detection_date + pd.to_timedelta(
    days_to_completion_gen3.values.ravel(), unit="D"
)
print(completion_dates_gen3)


# also extract the generation 3 value at the latitude and longiutude supplied in the attributes
lat = data.attrs["latitude"]
lon = data.attrs["longitude"]
gen3_at_location = data["days_to_completion"].sel(
    generation=3, latitude=lat, longitude=lon, method="nearest"
)
print(
    f"Generation 3 days to completion at location ({lat}, {lon}): {gen3_at_location.values}"
)
# date
print(
    "date of completion:",
    detection_date + pd.to_timedelta(gen3_at_location.values, unit="D"),
)
```

```python
import geoviews as gv
import holoviews as hv
import xarray as xr
import numpy as np

# Enable geoviews and holoviews extensions
gv.extension("bokeh")
hv.extension("bokeh")

# Load the dataset
data = xr.open_dataset("santaclara_testing_off_results.nc")

# Extract the variable to plot
days_to_completion = data["days_to_completion"].sel(generation=3)

# Get metadata for the plot
latitude = data.coords["latitude"].values
longitude = data.coords["longitude"].values
global_min = float(days_to_completion.min())
global_max = float(days_to_completion.max())

# Create the plot
tiles = gv.tile_sources.OSM.opts(alpha=1.0, bgcolor="white")
img = gv.Image(
    (longitude, latitude, days_to_completion.values),
    kdims=["longitude", "latitude"],
    vdims=["Days to Completion"],
).opts(
    cmap="Viridis",
    colorbar=True,
    clim=(global_min, global_max),
    tools=["hover"],
    width=800,
    height=600,
)

# Combine the OSM tiles with the data plot
plot = tiles * img

# Save the plot to a PNG file
hv.save(plot, "days_to_completion_plot.png", fmt="png")u
```
